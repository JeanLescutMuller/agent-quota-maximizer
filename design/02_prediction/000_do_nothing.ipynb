# 000 · do nothing — the real null hypothesis

**Reserve the whole window for the human, always.** The bot is never allowed to spend
anything, so every percent the human does not use simply expires.

This is not a forecaster. It is **the world before this project existed**, and it is
the number every other approach has to beat to justify the project at all:

```
  reserve = 100%   ->   stolen = 0 by construction
                        wasted = everything the human did not use
```

It is deliberately **not tuned**. Tuning it would turn it into `00_constant`, which is
a different question (the best fixed reserve). This one is fixed at 100 because that
is what doing nothing costs.

````python
import sys, pathlib, warnings
warnings.filterwarnings("ignore")
sys.path.insert(0, str(pathlib.Path.cwd() / "lab"))
import numpy as np, pandas as pd
import dataset as D

frame = D.build("claude")
train, validation, test = D.split(frame)
COLS = D.features(frame)
print("\ncost model:  stolen $%.2f/pct   wasted $%.2f/pct   ratio %.0f:1"
      % (D.COST_STOLEN, D.COST_WASTED, D.COST_STOLEN / D.COST_WASTED))
print("the cost-optimal quantile is therefore %.4f" % D.optimal_quantile())
````

````python
val = [D.evaluate("do nothing", validation["y%d" % N].values,
                  np.full(len(validation), 100.0)) | {"N": N} for N in D.HORIZONS]
print(pd.DataFrame(val).set_index("N")[["cost_usd", "stolen", "wasted", "coverage"]].round(3))
print()
print("stolen is exactly 0 at every horizon, by construction: the bot never runs.")
print("wasted is the whole point of the project, measured.")
D.save("000_do_nothing", {"kind": "constant",
                          "per_horizon": {str(N): {"pct": 100.0} for N in D.HORIZONS}}, val)
````
